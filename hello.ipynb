OK, now let's analyze that statement:

print("Hello Python Universe! \n")


**_print_** is a built-in function in python to print text to the screen.  
_"Hello Python Universe \n"_ is a **string** in python, we enclose strings in quotes - single or double quotes both work.
**_\n_** is a special character in python that says to end the line (add a blank line)

```python
print(test)
```

That didn't work because **test is not a string**, I needed to use "test" it I wanted to print the string 'test'.  As is, python expects *test* to be a variable that was defined earlier in the code.  It wasn't defined, so python tells us that in the error.  

```python
print("test")
```

much better...

```python
print("test \n\n\n\n\n\n\n")
```

See all of the blank lines?   We can also enclose strings in single quotes:

```python
print('test\n')
```

See that works too!
